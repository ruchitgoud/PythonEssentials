**Loops, Lists, Tuples & Dictionaries**

```python
arr = [1, 1, 2, 1, 3, 4, 1, 3, 4]
freqCalc = {}

for i in range(len(arr)):

    if(arr[i] in freqCalc):
        freqCalc.update({arr[i]: freqCalc[arr[i]] + 1})

    else:
        freqCalc.update({arr[i]: 1})

print(freqCalc)
```

```python
numbers = list(map(int, input("Enter numbers: ").split()))

total = 0

for num in numbers:
    total += num

print(f"Numbers: {numbers}")
print(f"Sum: {total}")
print(f"Largest: {max(numbers)}")
print(f"Smallest: {min(numbers)}")
print(f"Average: {total / len(numbers):.2f}")
```
