**Sets, Functions, Scope & Recursion**

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)


num = int(input("Enter a number: "))

print(f"Factorial of {num} = {factorial(num)}")
```
